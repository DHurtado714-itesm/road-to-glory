## Strings

This is a sheat sheet to help you with strings in Python for Technical Interviews.

```python
string = "Hello, World!"

# Get the first character of the string
first_char = string[0]

# Get the last character of the string
last_char = string[-1]

# Get the first 5 characters of the string
first_five_chars = string[:5]

# Get the last 3 characters of the string
last_three_chars = string[-3:]

# Get the characters from the 3rd to the 5th character of the string
chars_3_to_5 = string[2:5]

# Get the length of the string
length = len(string)

# Get the index of the first occurrence of the character 'o'
index_of_o = string.index('o')

# Get the number of occurrences of the character 'o'
number_of_o = string.count('o')

# Check if the string starts with 'Hello'
starts_with_hello = string.startswith('Hello')

# Check if the string ends with 'World!'
ends_with_world = string.endswith('World!')

# Replace 'Hello' with 'Hi'
replaced = string.replace('Hello', 'Hi')

# Convert the string to uppercase
uppercase = string.upper()

# Convert the string to lowercase
lowercase = string.lower()

# Append at the end of the string
string += ' Goodbye!'

# Prepend at the beginning of the string
string = 'Hi, ' + string

# Split the string by ','
splitted = string.split(',')

# Join the list of strings with ','
joined = ','.join(splitted)
```
