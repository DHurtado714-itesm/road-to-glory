## Dictionaries in Python

Dictionaries are a fundamental data structure in Python, used to store key-value pairs. They are highly efficient for lookups, insertions, and deletions. Here are some key points about dictionaries:

- **Creation**: Dictionaries can be created using curly braces `{}` with key-value pairs separated by colons, or by using the `dict()` function.
- **Accessing Values**: Values can be accessed using their corresponding keys.
- **Modifying Values**: Values can be updated by assigning a new value to an existing key.
- **Adding Items**: New key-value pairs can be added by assigning a value to a new key.
- **Removing Items**: Items can be removed using the `del` statement or the `pop()` method.
- **Iterating**: You can iterate over keys, values, or key-value pairs using loops.

Dictionaries are widely used in technical interviews to test a candidate's understanding of data structures and their ability to manipulate data efficiently.

```python
# Example 1: Creating a dictionary to store student grades
grades = {
  'Alice': 85,
  'Bob': 92,
  'Charlie': 78
}
print("Student Grades:", grades)

# Example 2: Accessing a value using a key
alice_grade = grades['Alice']
print("Alice's Grade:", alice_grade) # Output: 85

# Example 3: Updating a value
grades['Alice'] = 88
print("Updated Grades:", grades)

# Example 4: Adding a new key-value pair
grades['David'] = 90
print("Grades after adding David:", grades)

# Example 5: Removing a key-value pair
del grades['Charlie']
print("Grades after removing Charlie:", grades)

# Using the get() method
grade = grades.get('Alice')
print("Alice's Grade:", grade) # Output: 88

# Using the get() method with a default value
grade = grades.get('Charlie', 0) # Output: 0

# Using the keys() method
students = grades.keys()
print("Students:", students) # Output: dict_keys(['Alice', 'Bob', 'David'])

# Using the values() method
grades_list = grades.values()
print("Grades:", grades_list) # Output: dict_values([88, 92, 90])

# Using the items() method
student_grades = grades.items()
print("Student Grades:", student_grades) # Output: dict_items([('Alice', 88), ('Bob', 92), ('David', 90)])

# Updating a dictionary using the update() method
grades.update({'Alice': 85, 'Bob': 92, 'Charlie': 78})
print("Updated Grades:", grades) # Output: {'Alice': 85, 'Bob': 92, 'Charlie': 78}

# Example 6: Iterating over keys and values
for student, grade in grades.items():
  print(f"{student}: {grade}")
```
