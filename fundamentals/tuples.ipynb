## Tuples

Tuples are immutable and ordered collections. Once created, they cannot be modified.

```python
# Creating a tuple
coordinates = (10, 20)

# Tuples are immutable
#coordinates[0] = 20 # This will raise an error

# Accessing elements
x, y = coordinates

# Tuples can be used as dictionary keys
location = {coordinates: "Home"}
print(location)
```
