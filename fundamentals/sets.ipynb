## Sets

Sets are unordered collections of unique items. Sets are used to store multiple items in a single variable. Sets are written with curly brackets.

```python
# Creating a set
unique_numbers = {1, 2, 3, 4, 4}

# Adding an item
unique_numbers.add(5)

# Removing an item
unique_numbers.remove(2)

# Set operations
a = {1, 2, 3}
b = {3, 4, 5}
union = a | b  # {1, 2, 3, 4, 5}
intersection = a & b  # {3}
```
