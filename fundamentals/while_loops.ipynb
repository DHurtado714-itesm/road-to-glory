## While Loops

### Description
A `while` loop is a control flow statement that allows code to be executed repeatedly based on a given Boolean condition. The `while` loop can be thought of as a repeating `if` statement. The code inside the loop will continue to execute as long as the condition remains `True`.

### Usage in Tech Interviews
In technical interviews, `while` loops are often used to solve problems that require repeated execution of a block of code until a certain condition is met. They are particularly useful for scenarios where the number of iterations is not known beforehand, such as reading data until the end of a file, processing user input until a valid response is received, or implementing algorithms that converge to a solution iteratively.

### Example
Here is a simple example of a `while` loop that prints numbers from 1 to 5:

```python
i = 1
while i <= 5:
  print(i)
  i += 1
```


````python
# Example 1: Reading user input until a valid response is received
response = ""
while response.lower() not in ["yes", "no"]:
  response = input("Do you want to continue? (yes/no): ")
print(f"Your response: {response}")

# Example 2: Finding the first number divisible by both 3 and 5
num = 1
while True:
  if num % 3 == 0 and num % 5 == 0:
    break
  num += 1
print(f"The first number divisible by both 3 and 5 is: {num}")

# Example 3: Summing numbers until the sum exceeds 100
total = 0
i = 1
while total <= 100:
  total += i
  i += 1
print(f"The sum exceeded 100 at: {total}")
````
